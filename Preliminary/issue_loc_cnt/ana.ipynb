```python
import json
result_path = "/home/jiawei/RepoCodeLoc/tools/Preliminary/issue_loc_cnt/sympy.jsonl"
result = []
with open(result_path, "r") as f:
    for line in f:
        result.append(json.loads(line))
    

cnt_have_related_loc = 0
cnt_not_have_related_loc = 0

for item in result:
    if item['have_related_loc'] is True:
        cnt_have_related_loc += 1
    elif item['have_related_loc'] is False:
        cnt_not_have_related_loc += 1
    else:
        raise ValueError("Invalid value for have_related_loc")

print(f"total_counts={len(result)}")
print(f"{cnt_not_have_related_loc=}")
print(f"{cnt_have_related_loc=}")
```
